```python
import matplotlib.pyplot as plt
import matplotlib.colors as colors
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from cartopy.util import add_cyclic_point
from cartopy.mpl.ticker import LongitudeFormatter, LatitudeFormatter


def plot_map(
    data,
    lat,
    lon,
    title,
    col_label,
    lat_bounds=None,
    lon_bounds=None,
    colormap="RdBu_r",
    centered_norm=True,
    central_lon=180,
    add_cyclic="auto",
    colorbar_shrink=0.7,
    lon_tick_spacing=30,
    lat_tick_spacing=15,
    vmin=None,
    vmax=None,
    extend='both'
):
    """
    Plot a lat/lon field using pcolormesh.

    Parameters
    ----------
    data : array-like
        2D field with shape (lat, lon).
    lat, lon : array-like
        1D latitude and longitude coordinates.
    title : str
        Plot title.
    col_label : str
        Colorbar label.
    lat_bounds : tuple or None
        Latitude plotting bounds, e.g. (-60, 60).
        If None, use the full latitude range.
    lon_bounds : tuple or None
        Longitude plotting bounds, e.g. (0, 360) or (-180, 180).
        If None, use the full longitude range.
    colormap : str or Colormap
        Colormap name or object.
    centered_norm : bool
        If True, center the colormap normalization around zero.
    central_lon : float
        Central longitude of the map projection.
    add_cyclic : {"auto", True, False}
        Whether to add a cyclic longitude point.
    colorbar_shrink : float
        Colorbar shrink factor.
    lon_tick_spacing : float
        Longitude tick spacing in degrees.
    lat_tick_spacing : float
        Latitude tick spacing in degrees.
    vmin, vmax : float or None
        Optional color scale limits.
    """
    data = np.asarray(data)
    lat = np.asarray(lat)
    lon = np.asarray(lon)

    def _needs_cyclic(lon_vals):
        if lon_vals.ndim != 1 or lon_vals.size < 2:
            return False

        lon_vals = np.asarray(lon_vals, dtype=float)
        lon_span = np.nanmax(lon_vals) - np.nanmin(lon_vals)
        step = np.nanmedian(np.abs(np.diff(lon_vals)))

        already_cyclic = np.isclose(
            abs(lon_vals[-1] - lon_vals[0]),
            360.0,
            atol=step * 1.5,
        )

        return lon_span >= 300.0 and not already_cyclic

    # Add cyclic point for global data if necessary
    if add_cyclic is True or (
        add_cyclic == "auto" and _needs_cyclic(lon)
    ):
        data, lon = add_cyclic_point(data, coord=lon)

    # Projection
    projection = ccrs.PlateCarree(
        central_longitude=central_lon
    )

    fig, ax = plt.subplots(
        figsize=(16, 6),
        subplot_kw={"projection": projection},
    )

    # Colour normalization
    if centered_norm:
        if vmin is not None and vmax is not None:
            halfrange = max(abs(vmin), abs(vmax))
            norm = colors.CenteredNorm(
                vcenter=0,
                halfrange=halfrange,
            )
        else:
            norm = colors.CenteredNorm(vcenter=0)
    else:
        norm = colors.Normalize(
            vmin=vmin,
            vmax=vmax,
        )

    # Plot individual grid cells
    mesh = ax.pcolormesh(
        lon,
        lat,
        data,
        cmap=colormap,
        norm=norm,
        shading="auto",
        transform=ccrs.PlateCarree(),
    )

    # Map bounds
    if lon_bounds is None:
        lon_bounds = (
            float(np.nanmin(lon)),
            float(np.nanmax(lon)),
        )

    if lat_bounds is None:
        lat_bounds = (
            float(np.nanmin(lat)),
            float(np.nanmax(lat)),
        )

    ax.set_extent(
        [
            lon_bounds[0],
            lon_bounds[1],
            lat_bounds[0],
            lat_bounds[1],
        ],
        crs=ccrs.PlateCarree(),
    )

    # Coastlines / borders
    ax.coastlines()
    ax.add_feature(cfeature.BORDERS)

    # Latitude / longitude ticks
    lon_ticks = np.arange(
        np.ceil(lon_bounds[0] / lon_tick_spacing) * lon_tick_spacing,
        lon_bounds[1] + lon_tick_spacing,
        lon_tick_spacing,
    )

    lat_ticks = np.arange(
        np.ceil(lat_bounds[0] / lat_tick_spacing) * lat_tick_spacing,
        lat_bounds[1] + lat_tick_spacing,
        lat_tick_spacing,
    )

    ax.set_xticks(
        lon_ticks,
        crs=ccrs.PlateCarree(),
    )
    ax.set_yticks(
        lat_ticks,
        crs=ccrs.PlateCarree(),
    )

    ax.xaxis.set_major_formatter(
        LongitudeFormatter()
    )
    ax.yaxis.set_major_formatter(
        LatitudeFormatter()
    )

    # Optional grid lines
    ax.gridlines(
        crs=ccrs.PlateCarree(),
        linewidth=0.5,
        alpha=0.4,
        linestyle="--",
    )

    ax.set_title(title)

    # Colorbar
    cbar = fig.colorbar(
        mesh,
        ax=ax,
        orientation="horizontal",
        shrink=colorbar_shrink,
        pad=0.10,
        extend=extend
    )
    cbar.set_label(col_label)

    plt.show()
```

```python
fcst_precip = xr.open_mfdataset(fcst_files).precipitation.load()
fcst_precip['time'] = fcst_precip.time + pd.Timedelta(1,'d')
fcst_precip
```

```python
mean_fcst_precip = fcst_precip.mean(dim="member")
```

```python
fcst_precip
```

```python
quantile = fcst_precip.quantile(0.8, dim="member")
prob_over_quantile = abs(fcst_precip - quantile).argmin(dim="member")/1000
```

```python
for t in mean_fcst_precip.time[:50]:
    plot_map(
        mean_fcst_precip.sel(time=t).squeeze(),
        diff.lat,
        diff.lon,
        f'cGAN ensemble mean precipitation\n{t.dt.date.values}',
        'mm/h',
        colormap='Blues',
        centered_norm=False,
        central_lon=0,
        lat_bounds=(50.5,52.5),
        lon_bounds=(-1.5,1.5),
        lon_tick_spacing=0.5,
        lat_tick_spacing=0.5,
        colorbar_shrink=0.35,
        vmin=0,
        vmax=0.1,
        extend='max'
    )
```

```python
for t in prob_over_quantile.time[:50]:
    plot_map(
        prob_over_quantile.sel(time=t).squeeze(),
        diff.lat,
        diff.lon,
        f'cGAN probability of precip > 80th percentile\n{t.dt.date.values}',
        'mm/h',
        colormap='Blues',
        centered_norm=False,
        central_lon=0,
        lat_bounds=(50.5,52.5),
        lon_bounds=(-1.5,1.5),
        lon_tick_spacing=0.5,
        lat_tick_spacing=0.5,
        colorbar_shrink=0.35,
        vmin=0,
        # vmax=40,
        extend='neither'
    )
```
